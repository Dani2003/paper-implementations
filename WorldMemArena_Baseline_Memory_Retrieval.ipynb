```python
# Step 5 — Build a simple retrieval baseline

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import re

def flatten_sessions(sample):
    records = []

    for session in sample["sessions"]:
        session_id = session["_v2_session_id"]

        for turn_idx, turn in enumerate(session["dialogue"]):
            content = turn.get("content", "")

            if content and content.strip():
                records.append({
                    "session_id": session_id,
                    "turn_idx": turn_idx,
                    "role": turn["role"],
                    "text": content.strip()
                })

    return records


records = flatten_sessions(sample)

print("Total dialogue records:", len(records))

# Build corpus
corpus = [r["text"] for r in records]

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=20000
)

X = vectorizer.fit_transform(corpus)

print("TF-IDF matrix:", X.shape)
```

```python
def retrieve(query, top_k=5):
    q_vec = vectorizer.transform([query])

    scores = cosine_similarity(q_vec, X).flatten()

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for idx in top_indices:
        results.append({
            "score": float(scores[idx]),
            "session_id": records[idx]["session_id"],
            "turn_idx": records[idx]["turn_idx"],
            "role": records[idx]["role"],
            "text": records[idx]["text"]
        })

    return results
```

```python
question = sample["qa_checkpoints"][0]["questions"][0]["question"]

print("QUESTION:")
print(question)

print("\nRETRIEVED CONTEXT:")
print("=" * 70)

results = retrieve(question, top_k=5)

for i, r in enumerate(results, 1):
    print(f"\n[{i}] score={r['score']:.4f}")
    print(f"session={r['session_id']} | role={r['role']}")
    print(r["text"][:700])
```

```python
def build_multimodal_records(sample):
    records = []

    for session in sample["sessions"]:
        session_id = session["_v2_session_id"]

        for turn_idx, turn in enumerate(session["dialogue"]):
            # Normal dialogue text
            content = turn.get("content", "").strip()

            if content:
                records.append({
                    "session_id": session_id,
                    "turn_idx": turn_idx,
                    "type": "text",
                    "text": content
                })

            # Image captions
            for att in turn.get("attachments", []):
                caption = att.get("caption", "").strip()

                if caption:
                    records.append({
                        "session_id": session_id,
                        "turn_idx": turn_idx,
                        "type": "image_caption",
                        "image_id": att.get("image_id"),
                        "text": caption
                    })

    return records


mm_records = build_multimodal_records(sample)

print("Multimodal records:", len(mm_records))

for r in mm_records[:5]:
    print("\n", r)
```

```python
mm_corpus = [r["text"] for r in mm_records]

mm_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=20000
)

mm_X = mm_vectorizer.fit_transform(mm_corpus)

print("Matrix:", mm_X.shape)
```

```python
def retrieve_multimodal(query, top_k=5):
    q_vec = mm_vectorizer.transform([query])

    scores = cosine_similarity(q_vec, mm_X).flatten()
    top_indices = np.argsort(scores)[::-1][:top_k]

    return [
        {
            "score": float(scores[idx]),
            "session_id": mm_records[idx]["session_id"],
            "turn_idx": mm_records[idx]["turn_idx"],
            "type": mm_records[idx]["type"],
            "image_id": mm_records[idx].get("image_id"),
            "text": mm_records[idx]["text"]
        }
        for idx in top_indices
    ]
```

```python
# Step 11 — Memory-point retrieval baseline

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def extract_memory_points(sample):
    memories = []

    for mp_session in sample["memory_points"]:
        session_id = mp_session["session_id"]

        for mp in mp_session["memory_points"]:
            memories.append({
                "memory_id": mp["memory_id"],
                "session_id": session_id,
                "text": mp["memory_content"],
                "memory_type": mp.get("memory_type"),
                "is_update": mp.get("is_update"),
                "update_type": mp.get("update_type"),
            })

    return memories


memories = extract_memory_points(sample)

print("Memory points:", len(memories))

memory_corpus = [m["text"] for m in memories]

memory_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=20000
)

memory_X = memory_vectorizer.fit_transform(memory_corpus)

print("Memory TF-IDF matrix:", memory_X.shape)
```

```python
def retrieve_memories(query, top_k=5):
    q_vec = memory_vectorizer.transform([query])

    scores = cosine_similarity(q_vec, memory_X).flatten()
    indices = np.argsort(scores)[::-1][:top_k]

    return [
        {
            "rank": rank + 1,
            "score": float(scores[idx]),
            **memories[idx]
        }
        for rank, idx in enumerate(indices)
    ]
```

```python
question = q["question"]

results = retrieve_memories(question, top_k=5)

print("QUESTION:")
print(question)

print("\nRETRIEVED MEMORIES:")
print("=" * 70)

for r in results:
    print(f"\nRank {r['rank']} | score={r['score']:.4f}")
    print("Memory ID:", r["memory_id"])
    print("Session:", r["session_id"])
    print("Text:", r["text"])
```

```python
ground_truth_memory_ids = {
    e["memory_id"]
    for e in q["evidence"]
    if "memory_id" in e
}

retrieved_ids = [r["memory_id"] for r in results]

print("\nGround truth:", ground_truth_memory_ids)
print("Retrieved:", retrieved_ids)

print(
    "\nRecall@1:",
    bool(set(retrieved_ids[:1]) & ground_truth_memory_ids)
)

print(
    "Recall@5:",
    bool(set(retrieved_ids[:5]) & ground_truth_memory_ids)
)
```

```python
from collections import Counter, defaultdict
import os
import json
import numpy as np

# ---------------------------------------------------------
# Find all locally available WorldMemArena JSON samples
# ---------------------------------------------------------

available_samples = []

for root, dirs, files in os.walk(snapshot_dir):
    for filename in files:
        if filename.endswith(".json") and filename != "small_ids.json":
            path = os.path.join(root, filename)

            try:
                with open(path, "r") as f:
                    data = json.load(f)

                if "sample_id" in data and "qa_checkpoints" in data:
                    available_samples.append((data["sample_id"], path))
            except Exception:
                pass

print("Available benchmark samples:", len(available_samples))


# ---------------------------------------------------------
# Evaluate memory retrieval
# ---------------------------------------------------------

all_results = []

for sample_id, path in available_samples:

    with open(path, "r") as f:
        sample_data = json.load(f)

    memories = extract_memory_points(sample_data)

    if not memories:
        continue

    corpus = [m["text"] for m in memories]

    # Build an independent index for this sample
    vectorizer = TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2),
        max_features=20000
    )

    X = vectorizer.fit_transform(corpus)

    for checkpoint in sample_data.get("qa_checkpoints", []):

        for question in checkpoint.get("questions", []):

            # Only evaluate questions with memory evidence
            gt_memory_ids = {
                e["memory_id"]
                for e in question.get("evidence", [])
                if "memory_id" in e
            }

            if not gt_memory_ids:
                continue

            query = question["question"]

            q_vec = vectorizer.transform([query])

            scores = cosine_similarity(q_vec, X).flatten()

            top_indices = np.argsort(scores)[::-1][:5]

            retrieved_ids = [
                memories[i]["memory_id"]
                for i in top_indices
            ]

            hit_1 = bool(
                set(retrieved_ids[:1]) & gt_memory_ids
            )

            hit_5 = bool(
                set(retrieved_ids[:5]) & gt_memory_ids
            )

            all_results.append({
                "sample_id": sample_id,
                "checkpoint_id": checkpoint.get("checkpoint_id"),
                "question": query,
                "question_type": question.get(
                    "question_type_abbrev",
                    "UNKNOWN"
                ),
                "difficulty": question.get(
                    "difficulty",
                    "UNKNOWN"
                ),
                "gt_memory_ids": list(gt_memory_ids),
                "retrieved_ids": retrieved_ids,
                "recall_at_1": hit_1,
                "recall_at_5": hit_5
            })


print("\nEvaluation complete.")
print("Questions evaluated:", len(all_results))
```

```python
total = len(all_results)

r1 = sum(r["recall_at_1"] for r in all_results) / total
r5 = sum(r["recall_at_5"] for r in all_results) / total

print("=" * 60)
print("WORLDMEMARENA — TF-IDF MEMORY RETRIEVAL BASELINE")
print("=" * 60)

print(f"Samples evaluated : {len(set(r['sample_id'] for r in all_results))}")
print(f"Questions         : {total}")
print(f"Recall@1          : {r1:.4f} ({r1*100:.2f}%)")
print(f"Recall@5          : {r5:.4f} ({r5*100:.2f}%)")
```

```python
by_type = defaultdict(list)

for r in all_results:
    by_type[r["question_type"]].append(r)

print("\nBY QUESTION TYPE")
print("=" * 60)

for qtype, rows in sorted(by_type.items()):

    r1 = np.mean([r["recall_at_1"] for r in rows])
    r5 = np.mean([r["recall_at_5"] for r in rows])

    print(
        f"{qtype:8s} "
        f"N={len(rows):4d} "
        f"R@1={r1:.3f} "
        f"R@5={r5:.3f}"
    )
```
